3. **Kiểm định Giả thuyết Thống kê Phi tham số:**
   - `Mann-Whitney U Test`: Đo lường phần bù giá (**Rent Premium %**) của các tiện ích (máy lạnh, gác lửng, giờ tự do).
   - `Kruskal-Wallis ANOVA`: Khảo sát sự khác biệt đơn giá giữa các Quận và nhóm người đăng.
   - `Spearman Rank Correlation`: Mối tương quan cự ly và diện tích với đơn giá.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = Path('..').resolve()
sys.path.append(str(BASE_DIR))

from src.stats_tests import run_all_hypothesis_tests

PROCESSED_FULL = BASE_DIR / 'data' / 'processed' / 'rental_cleaned_full.csv'
df = pd.read_csv(PROCESSED_FULL)
print(f'Loaded cleaned dataset: {len(df):,} listings.')
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
```

## 1. Phân tích Không gian & Quy luật Suy giảm Khoảng cách (Distance Decay)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Distance to Q1 vs Price per m2
sns.regplot(data=df.sample(min(1000, len(df)), random_state=42), x='dist_to_q1_km', y='price_per_m2_k', 
            scatter_kws={'alpha': 0.35, 'color': '#1f77b4'}, line_kws={'color': 'red', 'linewidth': 2}, ax=axes[0])
axes[0].set_title('Quy luật suy giảm đơn giá thuê theo cự ly tới Quận 1', fontsize=11, fontweight='bold')
axes[0].set_xlabel('Khoảng cách tới Chợ Bến Thành, Q.1 (km)')
axes[0].set_ylabel('Đơn giá thuê (Nghìn VNĐ/m²)')

# Distance to VNU-HCM vs Rent Price
sns.regplot(data=df.sample(min(1000, len(df)), random_state=42), x='dist_to_vnuhcm_km', y='price_million',
            scatter_kws={'alpha': 0.35, 'color': '#2ca02c'}, line_kws={'color': 'orange', 'linewidth': 2}, ax=axes[1])
axes[1].set_title('Mối quan hệ giữa giá thuê và cự ly tới Khu Đô thị ĐHQG-HCM', fontsize=11, fontweight='bold')
axes[1].set_xlabel('Khoảng cách tới Làng Đại học ĐHQG-HCM (km)')
axes[1].set_ylabel('Giá thuê (Triệu VNĐ/tháng)')

plt.tight_layout()
plt.show()
```

## 2. Kiểm định Giả thuyết Thống kê (Mann-Whitney U, Kruskal-Wallis & Spearman)

```python
test_results = run_all_hypothesis_tests(df)

print('=== BẢNG KIỂM ĐỊNH TÁC ĐỘNG CỦA TIỆN ÍCH (Mann-Whitney U Test) ===')
display(test_results['amenity_tests'])

print('\n=== BẢNG TƯƠNG QUAN SPEARMAN VỚI ĐƠN GIÁ & CỰ LY ===')
display(test_results['spearman_correlations'])
```
